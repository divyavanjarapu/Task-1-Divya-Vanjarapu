```python
# Importing the libraries
import pandas as pd
import numpy as np
```

```python
# Load dataset
df = pd.read_excel("Dataset for Data Analytics.xlsx")
```

```python
#Data types
print("\nData Types:")
print(df.dtypes)
```

```python
#Identifying the missing values in data
print("Missing Values Before Cleaning:")
print(df.isnull().sum())
print("\nTotal Missing Values Before Cleaning:",df.isnull().sum().sum())
```

```python
# fill the Missing values 
numeric_cols = df.select_dtypes(include=np.number).columns

for col in numeric_cols:
    if df[col].isnull().sum() > 0:
        df[col] = df[col].fillna(df[col].median())

categorical_cols = df.select_dtypes(include="object").columns

for col in categorical_cols:
    if df[col].isnull().sum() > 0 and not df[col].mode().empty:
        df[col] = df[col].fillna(df[col].mode()[0])
```

```python
# Remove duplicates
duplicates = df.duplicated().sum()
print(f"Duplicate Rows Found: {duplicates}")
df = df.drop_duplicates().reset_index(drop=True)
```

```python
# Auto detect date columns and convert the type object to date
for col in df.columns:
    if df[col].dtype == "object":

        converted = pd.to_datetime(df[col], errors="coerce")
        date_keywords = ["date", "time", "dob", "birth", "created"]
        if (
            converted.notna().sum() > len(df) * 0.8
            and any(word in col.lower() for word in date_keywords)
        ):
            df[col] = converted
```

```python
# Clean text columns
for col in df.select_dtypes(include="object").columns:
    df[col] = df[col].apply(
        lambda x: x.strip() if isinstance(x, str) else x
    )
```

```python
# Summary
print("After Cleaning The Dataset:")
print("Shape:", df.shape)
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])
print("Duplicates Removed:", duplicates)

print("\nMissing Values Remaining:")
print(df.isnull().sum())
print("\nTotal Missing Values:",df.isnull().sum().sum())
print("\nData Types:")
print(df.dtypes)
```

```python
# Save cleaned dataset
df.to_excel("Cleaned_Dataset.xlsx", index=False)
print("\nData Cleaning Completed Successfully!")
```
